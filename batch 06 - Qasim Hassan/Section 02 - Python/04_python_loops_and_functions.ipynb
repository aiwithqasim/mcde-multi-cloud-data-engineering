Given an unsorted list of integers, find the longest sequence of consecutive numbers. The numbers do not need to appear consecutively in the original list.

nums = [100, 4, 200, 1, 3, 2, 5, 8, 7, 6]

```python
nums = [100, 4, 200, 1, 3, 2, 5, 8, 7, 6]
nums.sort()
for i in range(min(nums)-1,max(nums)):
  if nums[i] == i+1:
    print(nums[i])
  else:
    break
```

```python
nums  = [100, 4, 200, 1, 3, 2, 5, 8, 7, 6]
nums.sort()
current = []
longest = []
for n in nums:
  if current == [] or n == current[-1] + 1:
    current.append(n)
  else:
    current = [n]
  if len(current) > len(longest):
    longest = current
print(longest)
```

Write a Python program that uses a while loop to separate even and odd numbers from a given list.

numbers = [12, 7, 9, 20, 15, 8, 6, 11, 14, 3]

```python
numbers = [12, 7, 9, 20, 15, 8, 6, 11, 14, 3]

i = 0
result_even = []
result_odd = []

while i < len(numbers):
    if numbers[i] % 2 == 0:
      result_even.append(numbers[i])
    else:
      result_odd.append(numbers[i])

    i += 1

print(result_even, result_odd)
```
