```python
class human():
  # adding class variable
  city = 'Karachi'
  country = 'Pakistan'

  def __init__(self, name, color, weight, height, gender):
    self.name = name
    self.color = color
    self.weight = weight
    self.height = height
    self.gender = gender

  ## adding a methods arguments
  def breathing(self, speed):
    print(f"{self.name} is breathing with {speed}")

  def sleeping(self, hours):
    print(f"{self.name} is sleeping from {hours} hours")

  def working(self):
    print(f"{self.name} is working")
```

```python
qasim = human("Qasim", 'brown', 70, 5.9, 'male')
ayesha = human("Ayesha", 'brown', 50, 5.2, 'female')
```

```python
# calling all arrtibutes and methods for qasim
print(qasim.name)
print(qasim.color)
print(qasim.weight)
print(qasim.height)
print(qasim.gender)
qasim.breathing('calmly')
qasim.sleeping(5)
qasim.working()

## calling class variables
print(qasim.city)
print(qasim.country)
```

```python
# Example of Inhertance

class parents():

  def __init__(self, eye_color, skin_color, blood_group):
    self.eye_color = eye_color
    self.skin_color = skin_color
    self.blood_group = blood_group

  def hobby(self, hobby):
    print(f"have {hobby} Hobby")
```

```python
class child(parents):

  def __init__(self, eye_color, skin_color, blood_group, name):
    super().__init__(eye_color, skin_color, blood_group)
    self.name = name

  def coding(self):
    print(f"{self.name} Know Programing")
```

```python
pobj = parents('Black', "brown", 'A+')

## calling attribute and method of parent class
print(pobj.blood_group)
pobj.hobby('Newspaper Reading')
```

```python
cobj = child('Brown', 'black', 'AB+', 'Ali')

## calling attribute and method of child class
print(cobj.blood_group)
cobj.hobby('Reading')
cobj.coding()
```

- class father
- class mother
- class child which will inherit both father and mother class

```python
class father():
  def __init__(self, eye_color):
    self.eye_color = eye_color

  def hobby(self, hobby):
    print(f"have {hobby} Hobby")

class mother():
  def __init__(self, skin_color):
    self.skin_color = skin_color

  def Occupation(self, Occupation):
    print(f"have {Occupation} Occupation")

class child(father, mother):
  def __init__(self, eye_color, skin_color, name):
    father.__init__(self, eye_color)
    mother.__init__(self, skin_color)
    self.name = name

  def coding(self):
    print(f"{self.name} Know Programing")
```

```python
f_obj = father('Black')
m_obj = mother('Brown')
c_obj = child('Gray', 'Brown', 'Ali')
```

```python
# father obj

print(f_obj.eye_color)
f_obj.hobby('Reading')
print('---------')

# mother obj

print(m_obj.skin_color)
m_obj.Occupation('Teacher')
print('---------')

# child obj

print(c_obj.eye_color)
print(c_obj.skin_color)
c_obj.coding()
```

Abstraction: `concept of hiding technicalities`

```python
class shapes(): # abstract class
  def __init__(self, name):
    self.name = name

  def area(self):
    pass

  def perimeter(self):
    pass
```

```python
class circle(shapes):
  def __init__(self, name, radius):
    super().__init__(name)
    self.radius = radius

  def area(self):
    print(f"Area of {self.name} is {3.14*self.radius**2}")

  def perimeter(self):
    print(f"Perimeter of {self.name} is {2*3.14*self.radius}")
```

image.png

```python
class triangle(shapes):
  def __init__(self, name, base, height, side):
    super().__init__(name)
    self.base = base
    self.height = height
    self.side = side

  def area(self):
    return (self.base * self.height)/2

  def perimeter(self):
    return self.side + self.side + self.side # assume sides are equal
```

```python
tri_obj = triangle('Triangle', 10, 10, 10)
print(tri_obj.name)
print(tri_obj.area())
print(tri_obj.perimeter())
```

Polymorphism: `different behavior of smae method in different classes`

```python
class Animal():
  def __init__(self, name):
    self.name = name

  def speak(self):
    print(f"{self.name} can speak")
```

```python
class Dog(Animal):
  def __init__(self, name):
    super().__init__(name)

  def speak(self): # overiding happened here
    print(f"{self.name} can bark")
```

```python
dog_obj = Dog('Dog')
dog_obj.speak()
```

```python
class Employee(): #Abstract class

  def __init__(self, name):
    self.name = name

  def calculate_salary(self):
    pass

  def employee_details(self):
    pass
```

```python
# FullTimeEmployee
class FullTimeEmployee(Employee):
  def __init__(self, name, salary):
    super().__init__(name)
    self.salary = salary

  def calculate_salary(self):
    return self.salary

  def employee_details(self):
    print(f"Fulltime employee and the salary is {self.salary}")

# PartTimeEmployee
class PartTimeEmployee(Employee):
  def __init__(self, name, hours, rate):
    super().__init__(name)
    self.hours = hours
    self.rate = rate

  def calculate_salary(self):
    return self.hours * self.rate

  def employee_details(self):
    print(f"Parttime employee and the salary is {self.calculate_salary()}")

  # ContractEmployee
class ContractEmployee(Employee):
  def __init__(self, name, contract_amount):
    super().__init__(name)
    self.contract_amount = contract_amount

  def calculate_salary(self):
    return  self.contract_amount

  def employee_details(self):
    print(f"Contract employee and the salary is {self.calculate_salary()}")
```

```python
fullTimeEmp = FullTimeEmployee("FullTime",50000)
fullTimeEmp.employee_details()

partTimeEmp = PartTimeEmployee("PartTime", 5, 100)
partTimeEmp.employee_details()

contractEmp = ContractEmployee("Contract", 100000)
contractEmp.employee_details()
```

```python
class Notification:
  def __init__(self, notification, message):
    self.notification = notification
    self.message = message
  def send_notification(self, notification, message):
    print(f"{notification} appeared. The message is: {message}")

class SMSNotification(Notification):
  def __init__(self, notification, message):
    super().__init__(notification, message)
  def send_notification(self):
    print(f"{self.notification} notification appeared!. The message is: {self.message}")
class PushNotification(Notification):
  def __init__(self, notification, message):
    super().__init__(notification, message)
  def send_notification(self):
    print(f"{self.notification} notification appeared!. The message is: {self.message}")

class EmailNotification(Notification):
  def __init__(self, notification, message):
    super().__init__(notification, message)
  def send_notification(self):
    print(f"{self.notification} notification appeared!. The message is: {self.message}")
```

Encapsualtion: `process of hiding etails`

```python
class saylani():

  def __init__(self, name, email, cnic):
    self.name = name # public
    self._email = email # protected
    self.__cnic = cnic # private

  def public_method(self):
    print("This is public method")

  def _protected_method(self):
    print("This is protected method")

  def __private_method(self):
    print("This is private method")
```

```python
obj = saylani('Qasim', 'qasim@gmail.com', '42121-123456-0')
```

```python
print(obj.name)
print(obj._email)
print(obj._saylani__cnic)
```

```python
obj.public_method()
obj._protected_method()
obj._saylani__private_method()
```

TASK:

- Use encapsulation for class i:e., `StudentLogin()`:
- Initialize two static but private attributes i:e., `__user_name`, `__password`
- Create one private methods i:e., name `__dbconectivity()` that will validate you username & password. if both are valid then should print "valid user" otherwise "not valid"
- Create one public methods i:e., update password & it should update your private password

```python
class StudentLogin():
  def __init__(self):
    self.__user_name = 'Username'
    self.__password = 'Password'

  def __dbconectivity(self, username, password):
    if self.__user_name == username and self.__password == password:
      print(f"Valid User")
    else:
      print(f"Invalid User")

  def update_password(self, username, password, new_pass):
    if self.__user_name == username and self.__password == password:
      self.__password = new_pass
      print(f"Password Updated...!!!")
    else:
      print(f"Invalid User")
```
