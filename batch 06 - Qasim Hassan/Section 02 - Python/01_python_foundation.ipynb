## String Formatting

````python
f_name = "Qasim"
l_name = "Hassan"
print("My name is", f_name, l_name) # method - 0
````

````python
print("My name is "+ f_name + ' ' + l_name) # method -1: using concatenation
````

## Conditional Statements

```
if <conidition>:
  <Body of if>
elif <second condition> # elif can be so many
  <body of elif>
else:
  <Body of Else>
```



````python
age = int(input("Enter your age: "))

if age >= 18:
  print("Eligible for CNIC")
else:
  print("Not Eligible for CNIC")
````

````python
is_holiday = input("Is today Holiday: ")

print(is_holiday)
print(type(is_holiday))

if is_holiday == "True":
  print ("Today is Off")
elif is_holiday == "False":
  print("Today is On")
else:
  print("Invalid Input")
````
