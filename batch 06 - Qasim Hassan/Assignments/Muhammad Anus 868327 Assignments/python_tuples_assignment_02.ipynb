```python
#Q.no.17 # Q17. Nested Tuple Access

students = (
    ("Ali", (85, 90, 88)),
    ("Sara", (92, 87, 95)),
    ("Ahmed", (78, 84, 80))
)
sara_second = students[1][1][1]
ali_average = sum(students[0][1]) / len(students[0][1])
ahmed_highest = max(students[2][1])

print("Sara Second Score:", sara_second)
print("Ali Average:", ali_average)
print("Ahmed Highest Score:", ahmed_highest)
```

```python
#Q.no.18 Question: Sort by salary descending 
# without loops, comprehensions, or lambda.
import builtins
import operator

employees = (
    ("Ali", 80000),
    ("Sara", 120000),
    ("Ahmed", 95000)
)

sorted_employees = builtins.tuple(
    sorted(employees, key=operator.itemgetter(1), reverse=True)
)

print(sorted_employees)
```

```python
#Q19. Tuple Matrix
#Question: Extract diagonal, 
#calculate its product, and build a tuple of row sums.
matrix = ((2, 4, 6), (8, 10, 12), (14, 16, 18))

diagonal = (matrix[0][0], matrix[1][1], matrix[2][2])
product = diagonal[0] * diagonal[1] * diagonal[2]

row_sums = (
    sum(matrix[0]),
    sum(matrix[1]),
    sum(matrix[2])
)

print("Diagonal:", diagonal)
print("Product:", product)
print("Row sums:", row_sums)
```

```python
#Q20. Inventory Tuple Challenge
#Question: Calculate inventory value, find the most 
#expensive product, and build a reversed tuple of names.
records = (
    ("Laptop", 120000, 5),
    ("Mouse", 1500, 20),
    ("Keyboard", 3500, 10)
)

inventory_value = (
    records[0][1] * records[0][2]
    + records[1][1] * records[1][2]
    + records[2][1] * records[2][2]
)

most_expensive = max(records, key=__import__('operator').itemgetter(1))

reversed_names = (
    records[2][0],
    records[1][0],
    records[0][0]
)

print("Inventory value:", inventory_value)
print("Most expensive product:", most_expensive[0])
print("Reversed names:", reversed_names)
```
