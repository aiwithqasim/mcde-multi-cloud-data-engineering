```python
#Acessing the Elements
#Q.no.1: Extract the first middle and last elements without modifying list.
number = [10,20,30,40,50,60,70]
first_number = number [0]
mid_number = number[3]
last_number = number[6]
print("first_number",first_number)
print("middle",mid_number)
print("last_num",last_number)
```

```python
# Q.no.2 : Create three new lists:first five elements,last four elements,
#  and every second element in reverse order.
numbers = [5,10,15,20,25,30,35,40,45,50]
first_five = numbers[:5]
last_four = numbers[-4:]
reverse_every_second = numbers[::-2]
print("first five elements:", first_five)
print("last_four:", last_four)
print("Every second element in reverse order:", reverse_every_second)
```

```python
#Q.no.4 : Reverse usig slicing only
numbers = [12,45,23,67,89,34]
reverse_numbers = numbers[::-1]
print(reverse_numbers)
```

```python
#Q.no.5: Find the second largest unique number expected 67 
numbers = [45,12,89,67,45,23]
unique_numbers = list(set(numbers))
unique_numbers.sort()

second_largest = unique_numbers[-2]

print("Second largest unique number:", second_largest)
```

```python
#Q.no.6 : Rotate the list right by k using 
# slicing.Expected: [50,60,10,20,30,40]
numbers = [10,20,30,40,50,60]
k = 2
rotated_numbers = numbers[-k:] + numbers[:-k]
print(rotated_numbers)
```

```python
#Q.no.9  Extract Data from a Matrix

matrix = [
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
]

# Extract main diagonal
main_diagonal = [matrix[0][0], matrix[1][1], matrix[2][2]]

# Extract secondary diagonal
secondary_diagonal = [matrix[0][2], matrix[1][1], matrix[2][0]]

# Extract middle column
middle_column = [matrix[0][1], matrix[1][1], matrix[2][1]]

# Calculate sums
main_sum = sum(main_diagonal)
secondary_sum = sum(secondary_diagonal)
middle_sum = sum(middle_column)

print("Main diagonal:", main_diagonal, "Sum:", main_sum)
print("Secondary diagonal:", secondary_diagonal, "Sum:", secondary_sum)
print("Middle column:", middle_column, "Sum:", middle_sum)
```

```python
#Q.no.10 Build a list of the three largest unique values ascending 
# then two smallest unique values descending Expected [25,30,35,10,5]
data = [15,5,25,10,35,20,30]
unique_data = sorted(set(data))

largest_three = unique_data[-3:]
smallest_two = unique_data[:2][::-1]

result = largest_three + smallest_two

print(result)
```
