```python
#Q.no.24 Find students taking both and exactly one course.
python_students = {"Ali","sara","Ahmad","zain"}
sql_students = {"sara","Ahmad","Hamza","usman"}

both_courses = python_students & sql_students
exactly_one_course = python_students ^ sql_students

print("TALKING BOTH COURSES",sorted(both_courses))
print("exactly one courses", sorted(exactly_one_course))
```

```python
#Q.no27 Find students in al three ,exactly python and
# sql not aws and atleast two.
python = {"Ali","Ahmad","sara","zain"}
sql = {"sara","Ahmad","Hamza"}
aws = {"Ahmad","zain","Hamza"}
all_three = python & sql & aws
python_sql_notaws = (python & sql)-aws
atleast_two = (python & sql) | (python & aws) | (sql & aws)
print("All three elements:",all_three)
print("python_sql_notaws:", python_sql_notaws)
print("atleast_two elements:",atleast_two)
```

```python
#Q.no.28 Find IDs appearing on all days ,exactly one day and at least two days
day1 = {101,102,103,104}
day2 = {103,104,105,106}
day3 = {104,106,107,108}
all_days = day1 & day2 & day3
exactly_one_day = (day1 | day2 | day3) -(
   (day1 & day2)|
   (day1 & day3)|
   (day2 & day3)
)
at_least_two_days = (
    (day1 & day2)|
   (day1 & day3)|
   (day2 & day3)
)
print("Exactly one day:",sorted(exactly_one_day))
print("At least two days:", sorted(at_least_two_days))
```

```python
#Q.no.29 Find missing and extra permissions and test full authorization.
required_permissions = {"read","write","execute"}
user_permissions = {"read","write","delete"}

missing_permissions = required_permissions - user_permissions
extra_permissions = user_permissions - required_permissions
full_authorization = required_permissions.issubset(user_permissions)
print("Missing permissions:", missing_permissions)
print("Extra permissions:", extra_permissions)
print("Full authorization:", full_authorization)
```

```python
#Q.no.30 Find customers on exactly one,exactly two and three platforms.
website = {"c1","c2","c3","c4","c5"}
mobile = {"c3","c4","c6","c7"}
store = {"c2","c4","c7","c8"}

exactly_one = (website | mobile | store) - (
    (website & mobile) | (website & store) | (mobile & store)
)
exactly_two = (
    ((website & mobile) | (website & store) | (mobile & store))
    - (website & mobile & store)
)
three = website & mobile & store 
print("Exactly one platform: ",exactly_one)
print("Exactly two platforms:",exactly_two)
print("three platforms:",three)
```
