```python
#Q.no.35 Find highest lowest average and top-scoring students 
# without loops pr lambda
marks = {"Ali": 85,"Sara": 92,"Ahmed":78,"zain": 88}
highest_marks_student = max(marks.values())
lowest_marks = min(marks.values())
average = sum(marks.values()) / len(marks)
top_scoring_student = list(marks.keys())[list(marks.values()).index(highest_marks_student)]


print("Highest marks:", highest_marks_student)
print("Lowest marks:", lowest_marks)
print("Average marks:", average)
print("Top-scoring student:", top_scoring_student)
```

```python
#Q.no.38 Calcultate subtotals,grandtotal,and amount after 10% discount.
order = {
    "Laptop":{"price": 120000,"quantity":21},
    "mouse":{"price": 1500,"quantity":51},
    "keyboard":{"price":3500,"quantity":3}
}
laptop_subtotal = order["Laptop"]["price"] * order["Laptop"]["quantity"]
mouse_subtotal = order["mouse"]["price"] * order["mouse"]["quantity"]
keyboard_subtotal = order["keyboard"]["price"] * order["keyboard"]["quantity"]
grand_total = laptop_subtotal +mouse_subtotal + keyboard_subtotal
discount = grand_total * 0.10
final_amount = grand_total - discount
print("Laptop subtotal:", laptop_subtotal)
print("Mouse subtotal:", mouse_subtotal)
print("Keyboard subtotal:", keyboard_subtotal)
print("Grand total:", grand_total)
print("10% Discount:", discount)
print("Final amount:", final_amount)
```

```python
# Q.no.39: Calculate each average, top performer,
# and highest SQL score.

employees = {
    "Ali": {"Python": 90, "sql": 85, "aws": 88},
    "Sara": {"Python": 95, "sql": 92, "aws": 90},
    "Ahmed": {"python": 80, "sql": 87, "aws": 84}
}

ali_average = (90 + 85 + 88) / 3
sara_average = (95 + 92 + 90) / 3
ahmed_average = (80 + 87 + 84) / 3

print("Ali's average:", ali_average)
print("Sara's average:", sara_average)
print("Ahmed's average:", ahmed_average)

top_performer = max(
    employees,
    key=lambda name: sum(employees[name].values()) / 3
)
print("Top performer:", top_performer)

highest_sql_score = max(
    employees["Ali"]["sql"],
    employees["Sara"]["sql"],
    employees["Ahmed"]["sql"]
)
print("Highest SQL score:", highest_sql_score)
```

```python

#Q.no.40 1. Calculate the total value of the inventory.
# Find the most valuable item based on price × stock.
# Update the stock after selling two laptops and five mice.
inventory = {
    "P101": {"name": "Laptop", "price": 120000, "stock": 5},
    "P102": {"name": "Mouse", "price": 1500, "stock": 20},
    "P103": {"name": "Keyboard", "price": 3500, "stock": 10}
}

laptop_value = inventory["P101"]["price"] * inventory["P101"]["stock"]
mouse_value = inventory["P102"]["price"] * inventory["P102"]["stock"]
keyboard_value = inventory["P103"]["price"] * inventory["P103"]["stock"]

total_value = laptop_value + mouse_value + keyboard_value

print("Total inventory value:", total_value)

laptop_total = inventory["P101"]["price"] * inventory["P101"]["stock"]
mouse_total = inventory["P102"]["price"] * inventory["P102"]["stock"]
keyboard_total = inventory["P103"]["price"] * inventory["P103"]["stock"]

most_valuable_item = max(
    laptop_total, mouse_total, keyboard_total
)

if most_valuable_item == laptop_total:
    print("Most valuable item:", inventory["P101"]["name"])
elif most_valuable_item == mouse_total:
    print("Most valuable item:", inventory["P102"]["name"])
else:
    print("Most valuable item:", inventory["P103"]["name"])

inventory["P101"]["stock"] -= 2
inventory["P102"]["stock"] -= 5

print("Laptop stock after sale:", inventory["P101"]["stock"])
print("Mouse stock after sale:", inventory["P102"]["stock"])
print("Keyboard stock:", inventory["P103"]["stock"])
```
