```python
#Q.no.41 Calculate each average,top student tuple of names and set 
#of unique numbers.
students = [
    {"id": 1, "name": "Ali","marks":(85,90,88)},
    {"id": 2, "name": "Sara","marks":(92,95,91)},
    {"id": 3, "name": "Ahmed", "marks":(70,84,80)}
]
unique_numbers = set(
    students[0]["marks"]+
    students[1]["marks"]+
    students[2]["marks"]
)
average_student_1 = (85+90+88)/3
average_student_2 = (92+95+91)/3
average_student_3 = (70+84+80)/3
top_student = max(students, key=lambda student: sum(student["marks"]) / len(student["marks"]))
print("Average marks of Ali:",average_student_1)
print("Average marks of Sara:",average_student_2)
print("Average marks of Ahmed:",average_student_3)
print ("Top student Name:",top_student["name"])
print("of Marks:",top_student["marks"])
print("Unique numbers",unique_numbers)
```

```python
#Q.no.42 Find common skills Ali-only skills,all team skills,
# and employees meeting Python + SQL requirements.
employees = {
    "Ali": {"Python","SQL","aws"},
    "Sara": {"Python","Databricks","SQL"},
    "Ahmed": {"AWS","Snowflake","SQL"}
}
common_skills = (
    employees["Ali"]
    & employees["Sara"]
    & employees["Ahmed"]
)
ali_only = (
    employees["Ali"]
    - employees["Sara"]
    - employees["Ahmed"]
)
all_skills = (
    employees["Ali"]
    | employees["Sara"]
    | employees["Ahmed"]
)
required_skills = {"Python", "SQL"}

qualified_employees = {
    name for name, skills in employees.items()
    if required_skills.issubset(skills)
}

print("Common skills:", common_skills)
print("Ali-only skills:", ali_only)
print("All team skills:", all_skills)
print("Qualified employees:", qualified_employees)
```

```python
#Q.no.44 Calculate spending per customer, top customer, 
# and tuple of totals in customer ID order.

customers = {
    "C101": {"name": "Ali", "orders": [1200, 2500, 1800]},
    "C102": {"name": "Sara", "orders": [3000, 4500]},
    "C103": {"name": "Ahmed", "orders": [1500, 2000, 1000]}
}

# Calculate spending per customer
spending = {
    "C101": sum(customers["C101"]["orders"]),
    "C102": sum(customers["C102"]["orders"]),
    "C103": sum(customers["C103"]["orders"])
}

# Find top customer
top_customer = max(spending, key=spending.get)

# Tuple of totals in customer ID order
totals = (
    spending["C101"],
    spending["C102"],
    spending["C103"]
)

print("Spending per customer:", spending)
print("Top customer:", customers[top_customer]["name"])
print("Top customer spending:", spending[top_customer])
print("Tuple of totals:", totals)
```

```python
#Q.no.47 Find total processed records, failed pipelines, 
# largest pipeline total, and average records per run for each.

pipelines = {
    "pipeline_1": {"status": "success", "records": (1000, 1200, 1100)},
    "pipeline_2": {"status": "failed", "records": (800, 0, 0)},
    "pipeline_3": {"status": "success", "records": (1500, 1600, 1700)}
}
total_records = sum(pipelines["pipeline_1"]["records"]) + sum(pipelines["pipeline_2"]["records"]) + sum(pipelines["pipeline_3"]["records"])
failed_pipelines = {
    "pipeline_2"
}

pipeline_1_total = sum(pipelines["pipeline_1"]["records"])
pipeline_2_total = sum(pipelines["pipeline_2"]["records"])
pipeline_3_total = sum(pipelines["pipeline_3"]["records"])

largest_pipeline_total = max(
    pipeline_1_total, pipeline_2_total, pipeline_3_total
)

average_1 = pipeline_1_total / len(pipelines["pipeline_1"]["records"])
average_2 = pipeline_2_total / len(pipelines["pipeline_2"]["records"])
average_3 = pipeline_3_total / len(pipelines["pipeline_3"]["records"])

print("Total Processed Records:", total_records)
print("Failed Pipelines:", failed_pipelines)
print("Largest Pipeline Total:", largest_pipeline_total)
print("Pipeline 1 Average:", average_1)
print("Pipeline 2 Average:", average_2)
print("Pipeline 3 Average:", average_3)
```

```python
#Q.no.48 Assuming zero starting balances, calculate final balances, highest balance,
#  total withdrawals, and largest single deposit.

accounts = {
    "A101": {"holder": "Ali", "transactions": [5000, -2000, 3000, -1000]},
    "A102": {"holder": "Sara", "transactions": [10000, -3000, -2000]},
    "A103": {"holder": "Ahmed", "transactions": [7000, -1500, 2500]}
}
balance_ali = sum(accounts["A101"]["transactions"])
balance_sara = sum(accounts["A102"]["transactions"])
balance_ahmed = sum(accounts["A103"]["transactions"])

final_balances = {
    "Ali": balance_ali,
    "Sara": balance_sara,
    "Ahmed": balance_ahmed
}

highest_balance = max(
    balance_ali, balance_sara, balance_ahmed
)

total_withdrawals = 2000 + 1000 + 3000 + 2000 + 1500

largest_deposit = max(5000, 3000, 10000, 7000, 2500)

print("Final Balances:", final_balances)
print("Highest Balance:", highest_balance)
print("Total Withdrawals:", total_withdrawals)
print("Largest Single Deposit:", largest_deposit)
```

```python
#Q.no.49 Find products bought by at least two users, Ali wishlist minus purchases,
#  shared wishlist products, and all products.
users = {
    "Ali": {
        "purchased": {"Laptop", "Mouse"},
        "wishlist": {"Keyboard", "Monitor", "Mouse"}
    },
    "Sara": {
        "purchased": {"Keyboard", "Monitor"},
        "wishlist": {"Laptop", "Mouse"}
    },
    "Ahmed": {
        "purchased": {"Laptop", "Keyboard"},
        "wishlist": {"Monitor", "Headphones"}
    }
}
bought_by_two = (
    (users["Ali"]["purchased"] & users["Sara"]["purchased"])
    | (users["Ali"]["purchased"] & users["Ahmed"]["purchased"])
    | (users["Sara"]["purchased"] & users["Ahmed"]["purchased"])
)
ali_wishlist_only = (
    users["Ali"]["wishlist"] - users["Ali"]["purchased"]
)
shared_wishlist = (
    users["Ali"]["wishlist"]
    & users["Sara"]["wishlist"]
    & users["Ahmed"]["wishlist"]
)
all_products = (
    users["Ali"]["purchased"] | users["Ali"]["wishlist"]
    | users["Sara"]["purchased"] | users["Sara"]["wishlist"]
    | users["Ahmed"]["purchased"] | users["Ahmed"]["wishlist"]
)

print("Bought by at least two users:", bought_by_two)
print("Ali wishlist minus purchases:", ali_wishlist_only)
print("Shared wishlist products:", shared_wishlist)
print("All products:", all_products)
```
